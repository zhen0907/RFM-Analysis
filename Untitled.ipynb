```python
import pandas as pd

# 假設你的原始 DataFrame 名稱為 df
# 欄位包含：CustomerKey、Date、InvoNo、Amount

# 1. 確保Date欄位為 datetime 型態，以便進行Date相減
df['Date'] = pd.to_datetime(df['Date'])

# 2. 設定基準Date（即題目中的「Date欄最大Date」，通常是取整個資料集裡最新的那天作為參考點）
# 如果你是要跟「今天」比對，可以改成 pd.Timestamp('today')
reference_date = df['Date'].max()

# 3. 使用 groupby 進行聚合運算
df_rfm = (
    df.groupby('CustomerKey')
    .agg(
        F=('InvoNo', 'count'),  # InvoNo計數
        M=('Amount', 'sum'),  # Amount加總
        # 該客戶的最大Date 與 基準Date的差值（天數）
        # .dt.days 可以將時間差（Timedelta）轉換為整數天數
        R=('Date', lambda x: (reference_date - x.max()).days),
    )
    .reset_index()
)

# 4. 將CustomerKey欄位重新命名為 Customer
df_rfm = df_rfm.rename(columns={'CustomerKey': 'Customer'})

# 調整欄位順序為 Customer, F, M, R (非必要，但排版好看)
df_rfm = df_rfm[['Customer', 'F', 'M', 'R']]

# 檢視結果
print(df_rfm)
```

```python
df_rfm.to_csv('RFM_Data.csv',index=False)
```

```python
import pandas as pd

# 載入 RFM_Data.csv 檔案
df_rfm_loaded = pd.read_csv('RFM_Data.csv')

# 顯示前五筆資料
display(df_rfm_loaded.head())
```

```python
# 複製一份 df_rfm 進行正規化處理
nrfm = df_rfm.copy()

# 對 F, M, R 欄位進行 Min-Max 正規化
for col in ['F', 'M', 'R']:
    col_min = df_rfm[col].min()
    col_max = df_rfm[col].max()
    nrfm[col] = (df_rfm[col] - col_min) / (col_max - col_min)

# 檢視正規化後的資料
display(nrfm.head())
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns

# 提取用於分群的特徵欄位 (F, M, R)
features = nrfm[['F', 'M', 'R']]

# 計算不同 K 值的群內平方和 (Within-Cluster Sum of Squares, WCSS)
wcss = []
k_range = range(1, 11)
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init='auto')
    kmeans.fit(features)
    wcss.append(kmeans.inertia_)

# 繪製手肘法圖表 (Elbow Chart)
plt.figure(figsize=(8, 5))
plt.plot(k_range, wcss, marker='o', linestyle='--', color='b')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Inertia)')
plt.xticks(k_range)
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
# 執行 K-means 分群，群數設定為 4
kmeans_model = KMeans(n_clusters=4, random_state=42, n_init='auto')
nrfm['Cluster'] = kmeans_model.fit_predict(features)

# 計算各群的人數分佈
print("各群組人數分布：")
print(nrfm['Cluster'].value_counts())

# 檢視分群後的部分資料
display(nrfm.head())
```

```python
# 1. 將分群標籤回貼至客戶層級的 RFM 原始資料 df_rfm
df_rfm['Cluster'] = nrfm['Cluster']

# 2. 若您是指最原始的交易明細資料 df (包含 Date, InvoNo, CustomerKey, Amount)
# 我們可以透過 CustomerKey 與 Customer (df_rfm) 進行對應，將 Cluster 貼回去
df = df.merge(df_rfm[['Customer', 'Cluster']], left_on='CustomerKey', right_on='Customer', how='left')
df = df.drop(columns=['Customer'])  # 移除重複的對照欄位

# 顯示更新後的資料前五筆
print("客戶層級 RFM 資料 (df_rfm) 前五筆：")
display(df_rfm.head())

print("\n原始交易明細資料 (df) 前五筆：")
display(df.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 設定繪圖風格與字型以防中文亂碼
sns.set_theme(style="ticks")
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'Microsoft JhengHei', 'SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 繪製成對散布圖 (Pair Plot)，依 Cluster 欄位著色
# 將 Cluster 轉成字串/類別型態，讓調色盤能以離散方式呈現
plot_df = df_rfm.copy()
plot_df['Cluster'] = plot_df['Cluster'].astype(str)

g = sns.pairplot(
    plot_df[['R', 'F', 'M', 'Cluster']],
    hue='Cluster',
    palette='Set2',
    diag_kind='kde',
    plot_kws={'alpha': 0.6, 's': 30}
)

g.fig.suptitle('RFM 欄位成對散布圖 (Pair Plot) - 按 Cluster 分組', y=1.02, fontsize=14)
plt.show()
```

```python
# 定義 Cluster 與 Group 的對應關係
group_mapping = {
    0: 'Missing',
    1: 'High',
    2: 'Low',
    3: 'Medium'
}

# 在 df 中新增 Group 欄位
df['Group'] = df['Cluster'].map(group_mapping)

# 檢視前五筆更新後的資料，確保對應正確
display(df.head())
```

```python
# 1. 將 Group 欄位也對應到客戶層級的 df_rfm 資料集
df_rfm['Group'] = df_rfm['Cluster'].map(group_mapping)

# 2. 繪製以 Group 分組的成對散布圖 (Pair Plot)
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="ticks")
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'Microsoft JhengHei', 'SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 指定分類順序以確保圖例排序美觀 (High -> Medium -> Low -> Missing)
group_order = ['High', 'Medium', 'Low', 'Missing']

g = sns.pairplot(
    df_rfm[['R', 'F', 'M', 'Group']],
    hue='Group',
    hue_order=group_order,
    palette='Set1',
    diag_kind='kde',
    plot_kws={'alpha': 0.6, 's': 30}
)

g.fig.suptitle('RFM 欄位成對散布圖 (Pair Plot) - 按 Group 客群分組', y=1.02, fontsize=14)
plt.show()
```
